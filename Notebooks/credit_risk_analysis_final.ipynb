```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('credit_risk_dataset.csv')
```

```python
plt.figure(figsize=(8,6))

sns.countplot(
    data=df,
    x='loan_grade',
    hue='loan_status'
)

plt.title('Loan Status by Loan Grade')
plt.xlabel('Loan Grade')
plt.ylabel('Count')

plt.legend(title='Loan Status')
plt.show()
```

```python
default_rate = (
    df.groupby('loan_grade')['loan_status']
      .mean()
      .sort_index()
      * 100
)

plt.figure(figsize=(8,5))

default_rate.plot(
    kind='bar',
    color='#31572C'
)

plt.title('Default Rate by Loan Grade')
plt.xlabel('Loan Grade')
plt.ylabel('Default Rate (%)')

plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

```python
loan_intent_count = (
    df['loan_intent']
      .value_counts()
)

plt.figure(figsize=(9,6))

loan_intent_count.plot(
    kind='bar',
    color='#31572C'
)

plt.title('Loan Purpose Distribution')
plt.xlabel('Loan Purpose')
plt.ylabel('Number of Loans')

plt.xticks(rotation=25)

plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

```python
default_by_intent = (
    df.groupby('loan_intent')['loan_status']
      .mean()
)

default_by_intent = default_by_intent * 100

default_by_intent = default_by_intent.sort_values(ascending=False)

plt.figure(figsize=(8,5))

default_by_intent.plot(
    kind='bar',
    color='#31572C'
)

plt.title('Default Rate by Loan Purpose')
plt.xlabel('Loan Purpose')
plt.ylabel('Default Rate (%)')

plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("credit_risk_dataset.csv")
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    x='loan_status',
    y='loan_int_rate',
    data=df,
    color='#31572C'
)

plt.title('Interest Rate by Loan Status')
plt.xlabel('Loan Status')
plt.ylabel('Interest Rate (%)')

plt.show()
```

```python
interest_default = (
    df.groupby('loan_status')['loan_int_rate']
      .mean()
)

plt.figure(figsize=(6,5))

interest_default.plot(
    kind='bar',
    color='#31572C'
)

plt.title('Average Interest Rate by Loan Status')
plt.xlabel('Loan Status')
plt.ylabel('Average Interest Rate (%)')

plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```

```python
df['loan_status_label'] = df['loan_status'].map({
    0: 'No Default',
    1: 'Default'
})
```

```python
plt.figure(figsize=(7,5))

df['loan_status'].value_counts().sort_index().plot(
    kind='bar',
    color='#31572C'
)

plt.title('Loan Status Distribution')
plt.xlabel('Loan Status')
plt.ylabel('Count')

plt.xticks(
    ticks=[0,1],
    labels=['No Default', 'Default'],
    rotation=0
)

plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```

```python
average_income_percent = (
    df.groupby('loan_status')['loan_percent_income']
      .mean()
)

ax = average_income_percent.plot(
    kind='bar',
    figsize=(7,5),
    color='#31572C'
)

plt.title('Average Loan Percentage of Income by Loan Status')
plt.xlabel('Loan Status')
plt.ylabel('Average Loan Percent Income')

plt.xticks(
    ticks=[0,1],
    labels=['No Default', 'Default'],
    rotation=0
)

plt.grid(axis='y', linestyle='--', alpha=0.4)

for i, value in enumerate(average_income_percent):
    plt.text(i, value + 0.003, f'{value:.2f}', ha='center')

plt.show()
```
